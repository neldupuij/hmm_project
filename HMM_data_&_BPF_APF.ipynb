```python
def simulate_sv(n_obs, beta, phi, sigma, seed=0):
    """_summary_

    Args:
        n_obs (_type_): Number of observation
        beta (_type_): beta
        phi (_type_): phi
        sigma (_type_): sigma
        seed (int, optional): seed to have reproductible results

    Returns:
        _type_: generate data from the given parameters
    """
    rng = np.random.default_rng(seed)
    h = np.zeros(n_obs)
    h[0] = rng.normal(0, sigma / np.sqrt(1 - phi**2)) #Variance from Eq10 and the mean is 0 because we use from eq9 ℎ_𝑡 ⁡= 𝜙*⁢ℎ_𝑡−1 + 𝜂_𝑡
    for t in range(1, n_obs):
        h[t] = phi * h[t-1] + sigma * rng.normal() # AR(1) formula with 𝜂_𝑡 = N(0,1) 𝜂_𝑡 = σ_𝜂 * N(0,1)
    y = beta * np.exp(h / 2) * rng.normal(size=n_obs) # 𝑦𝑡⁢=𝛽⁢exp⁡(ℎ𝑡⁡/2)⁢𝜖𝑡  with 𝜖𝑡=N(0,1)
    return y, h
def systematic_resample(W, rng):
    N = len(W)
    positions = (rng.random() + np.arange(N)) / N
    cumsum = np.cumsum(W)
    cumsum[-1] = 1.0  # guard against floating-point rounding below 1
    return np.searchsorted(cumsum, positions)
#BPF, still need the APF
#Look at Appendix A.3 Algorithm A1 and 3.2.1 of the paper
def bootstrap_pf(y, beta, phi, sigma, N, c=2, seed=None):
    rng = np.random.default_rng(seed)
    h = rng.normal(0, sigma / np.sqrt(1 - phi**2), N) # (A1) Generate N particles from the stationary distribution, the same one used for h_1 in the simulator.
    log_W = np.full(N, -np.log(N)) # W_0^(i) = 1/N
    log_L_hat = 0.0
    for t in range(len(y)):
        if t > 0:
            h = phi * h + sigma * rng.normal(size=N) # (A2) Eq20 = N(φ * h_{t-1}^(i), σ_η²)
            
        log_w = (-0.5 * np.log(2 * np.pi) - np.log(beta) - h / 2
                - y[t]**2 / (2 * beta**2 * np.exp(h)))  # (A3) w_t^(i) = g * f / q, and with q = f this is w_t^(i) = g(y_t | h_t^(i)) where # y_t | h_t ~ N(0, β² * e^{h_t})
                                                        # log g(y_t | h_t) g is the density functino of yt|ht
                                                        
         # (A4) p_hat(y_t | y_{1:t-1}) = sum_i W_{t-1}^(i) * w_t^(i)
        log_Ww = log_w + log_W # w_t = omega_{t-1} * g
        m = log_Ww.max()
        log_p_hat = m + np.log(np.sum(np.exp(log_Ww - m)))
        log_L_hat += log_p_hat
        
        # (A5)  W_t^(i) proportional to W_{t-1}^(i) * w_t^(i), normalised to sum to 1
        log_W = log_Ww - log_p_hat
        W = np.exp(log_W)
        
        # (A6) resample if N_eff = 1 / sum_i (W_t^(i))^2 < N / c   Eq22
        N_eff = 1.0 / np.sum(W**2)
        if N_eff < N / c:                                     
            h = h[systematic_resample(W,rng)]
            log_W = np.full(N, -np.log(N)) # weights reset to 1/N
    return log_L_hat
```

```python
y, h = simulate_sv(n_obs=500, beta=0.9, phi=0.95, sigma=0.2, seed=1)

for N in [100, 1000, 10000]:
    loglik = [bootstrap_pf(y, 0.9, 0.95, 0.2, N=N, seed=s) for s in range(20)]
    print(N, np.mean(loglik), np.std(loglik))
```

So Above we simulate data that will then be use to see if our models work and retrieve the right parameters. To then be able to use this models on the real data to find the parameters ($\beta$, $\phi$ and $\sigma_\eta$)

```python
import numpy as np

#y_t | h_t ~ N(0, beta^2 * exp(h_t))
def log_obs_density(y_t, h, beta):
    """Log density of y_t | h_t ~ N(0, beta^2 * exp(h_t))"""
    return -0.5 * np.log(2.0 * np.pi) - np.log(beta) - 0.5 * h - 0.5 * (y_t / beta) ** 2 * np.exp(-h)


def log_normal_pdf(x, mean, std):
    """Log density of N(mean, std^2)"""
    return -0.5 * np.log(2.0 * np.pi) - np.log(std) - 0.5 * ((x - mean) / std) ** 2


def auxiliary_pf(y, beta, phi, sigma, N, c=2.0, seed=None):
    """
    Estimate log marginal likelihood log p(y_{0:T-1} | theta)
    """
    rng = np.random.default_rng(seed)
    n_obs = len(y)

    # Stationary initialization: h_0 ~ N(0, sigma^2 / (1 - phi^2))
    #Eq10 and (A1)
    h = rng.normal(0.0, sigma / np.sqrt(1.0 - phi**2), size=N) 
    log_W = np.full(N, -np.log(N))
    log_L_hat = 0.0

    for t in range(n_obs):
        if t == 0:
            # M_0 = P_0, so log(p_0 / m_0) = 0
            log_g = log_obs_density(y[0], h, beta)
            log_q = 0.0
            log_f= 0.0
        else:
            #Eq26 and also with the Eq9 parametrisation initialization
            # h_t = phi * h_{t-1} + eta_t with eta_t ~ N(0, sigma^2)
            # mu_t^(i) = E[h_t | h_{t-1}^(i)]
            #  = phi * h_{t-1}^(i) + E[eta_t]
            #  = phi * h_{t-1}^(i), since E[eta_t] = 0
            mu = phi * h
            
            #since q~N(mu_t1^(k) + (sigma^2 / 2) * ((y_t^2 / beta^2) * exp(-mu_t1^(k)) - 1), sigma^2)
            #we want the shift
            shift = 0.5 * sigma**2 * ((y[t]/beta) ** 2 * np.exp(-mu) - 1.0)
            total_shift=mu+shift
            h_new = rng.normal(total_shift, sigma) # (A2)

            log_g = log_obs_density(y[t], h_new, beta)
            log_f = log_normal_pdf(h_new, mu, sigma) # h_t|h_t-1 ~ N(phi*h_t-1,sigma^2)=N(mu,sigma^2)
            log_q = log_normal_pdf(h_new, total_shift, sigma) # h_t+1 | h_t^{k}, y_t+1; μ_t+1^{k} ~ N(shift,sigma^2)
            h = h_new

        # Potential w_t(h_{t-1}, h_t) # (A3)
        log_w = log_g + log_f - log_q

        # Incremental log-likelihood update
        #the equation of A4
        #log trick to not compute exponantials without under and overflows
        # a_i = log_Ww^(i)
        # log p_hat(y_t | y_[1:t-1], theta) = log(sum_{i=1_N} exp(a_i))
        # log(sum_{i=1_N} exp(a_i)) = log(exp(m) * sum_{i=1_N} exp(a_i_ - m)) = m + log(sum_{i=1_N} exp(a_i - m))
        log_Ww = log_W + log_w 
        max_log_Ww = np.max(log_Ww)
        log_p_hat = max_log_Ww + np.log(np.sum(np.exp(log_Ww - max_log_Ww)))
        log_L_hat += log_p_hat #the likelihood is the sum of the log_p_hat

        # Normalize weights
        # W_t^(i) ∝ W_t-1^(i) * w_t^(i)
        # but to make the ∝ an = sign that we need we know that to normalize we to divide by sum_{j=1_N} W_t-1^(j) * w_t^(j)
        log_W = log_Ww - log_p_hat
        W = np.exp(log_W)

        # Adaptive resampling
        if t < n_obs - 1:
            N_eff = 1.0 / np.sum(W**2)
            if N_eff < N / c:
                ancestors = systematic_resample(W, rng)
                h = h[ancestors]
                log_W = np.full(N, -np.log(N))

    return log_L_hat
```

```python
y, h = simulate_sv(n_obs=500, beta=0.9, phi=0.95, sigma=0.2, seed=1)

for N in [100, 1000, 10000]:
    loglik = [auxiliary_pf(y, 0.9, 0.95, 0.2, N=N, seed=s) for s in range(20)]
    print(N, np.mean(loglik), np.std(loglik))
```
